```python
import matplotlib.pyplot as plt

# Data
block_sizes = [4, 8, 16, 32]

# Full Search data
full_search_runtime = [0.753622, 0.200003, 0.051286, 0.0130214]

# Three Step Search data
tss_runtime = [0.110188, 0.0291636, 0.00839492, 0.00269596]

# Plotting
x = range(len(block_sizes))
width = 0.35

plt.figure(figsize=(10, 6))
plt.bar([i - width/2 for i in x], full_search_runtime, width=width, label="Full Search")
plt.bar([i + width/2 for i in x], tss_runtime, width=width, label="Three Step Search")

plt.xticks(x, block_sizes)
plt.xlabel("Block Size")
plt.ylabel("Runtime (seconds)")
plt.title("Runtime vs Block Size: Full Search vs Three Step Search")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Residual metric vs Search Area (Line Plot)
search_areas = [3, 5, 7, 9, 11]
tss_residual = [1.025954, 0.783066, 0.691721, 0.697927, 0.703192]
full_residual = [1.018870, 0.771169, 0.673921, 0.651008, 0.641415]

plt.figure(figsize=(10, 6))
plt.plot(search_areas, full_residual, marker='o', label="Full Search")
plt.plot(search_areas, tss_residual, marker='s', label="Three Step Search")

plt.xlabel("Search Area")
plt.ylabel("Residual Metric")
plt.title("Residual Metric vs Search Area")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Speedup vs Search Area (Bar Chart)
tss_speedup = [1.44, 3.49, 6.10, 9.78, 14.46]

plt.figure(figsize=(10, 6))
plt.bar(search_areas, tss_speedup, color='mediumseagreen')

plt.xlabel("Search Area")
plt.ylabel("Speedup (vs Full Search)")
plt.title("Three Step Search Speedup vs Full Search")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Performance Trade-off: Runtime vs Residual Metric (Scatter Plot)

# TSS data (runtime and residual metric for varying search areas)
tss_runtime_sa = [0.00852121, 0.00848046, 0.00839492, 0.00847733, 0.00810933]
# Full Search data
full_runtime_sa = [0.0123385, 0.0296802, 0.051286, 0.0829263, 0.117267]

plt.figure(figsize=(10, 6))
plt.scatter(full_residual, full_runtime_sa, color='royalblue', label="Full Search", s=100, marker='o')
plt.scatter(tss_residual, tss_runtime_sa, color='darkorange', label="Three Step Search", s=100, marker='s')

# Annotate best configurations
plt.annotate("Best Full Search", (0.641415, 0.117267), textcoords="offset points", xytext=(10, -10), ha='left')
plt.annotate("Best TSS", (0.703192, 0.00810933), textcoords="offset points", xytext=(-40, 10), ha='right')

plt.xlabel("Residual Metric (lower = better quality)")
plt.ylabel("Runtime (seconds, lower = faster)")
plt.title("Performance Trade-off: Runtime vs Residual Metric")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
