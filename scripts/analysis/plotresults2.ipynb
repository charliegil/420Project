```python
# Re-import matplotlib after kernel reset
import matplotlib.pyplot as plt

# === Data Setup === #

# Algorithms
algorithms = ['Full Search', 'Three Step Search', 'Diamond Search']

# Overall Comparison
avg_residual = [0.746581, 0.773534, 0.769918]
avg_runtime = [0.132311, 0.021503, 0.016838]

# Speedup vs Full Search
speedup_vs_full = [1, 6.15, 7.85]

# Block Sizes
block_sizes = [4, 8, 16, 32]
fs_runtime_block = [0.808848, 0.199986, 0.0562476, 0.0140597]
tss_runtime_block = [0.117186, 0.0307288, 0.0102598, 0.00319929]
ds_runtime_block = [0.0825117, 0.0254998, 0.00832537, 0.00365913]

# Residuals
fs_residual_block = [0.384974, 0.523103, 0.673921, 0.857045]
tss_residual_block = [0.428325, 0.536975, 0.691721, 0.849680]
ds_residual_block = [0.477561, 0.568641, 0.707417, 0.856095]

# === Visualization === #

# 1. Bar chart: Average Runtime & Residual Metric
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# Runtime
axs[0].bar(algorithms, avg_runtime, color=["royalblue", "darkorange", "mediumseagreen"])
axs[0].set_title("Average Runtime by Algorithm")
axs[0].set_ylabel("Runtime (s)")
axs[0].grid(True, linestyle='--', alpha=0.5)

# Residual Metric
axs[1].bar(algorithms, avg_residual, color=["royalblue", "darkorange", "mediumseagreen"])
axs[1].set_title("Average Residual Metric by Algorithm")
axs[1].set_ylabel("Residual Metric")
axs[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# 2. Line Plot: Runtime vs Block Size
plt.figure(figsize=(10, 6))
plt.plot(block_sizes, fs_runtime_block, marker='o', label="Full Search")
plt.plot(block_sizes, tss_runtime_block, marker='s', label="Three Step Search")
plt.plot(block_sizes, ds_runtime_block, marker='^', label="Diamond Search")
plt.xlabel("Block Size")
plt.ylabel("Runtime (s)")
plt.title("Runtime vs Block Size")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 3. Line Plot: Residual Metric vs Block Size
plt.figure(figsize=(10, 6))
plt.plot(block_sizes, fs_residual_block, marker='o', label="Full Search")
plt.plot(block_sizes, tss_residual_block, marker='s', label="Three Step Search")
plt.plot(block_sizes, ds_residual_block, marker='^', label="Diamond Search")
plt.xlabel("Block Size")
plt.ylabel("Residual Metric")
plt.title("Residual Metric vs Block Size")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 4. Speedup Comparison Bar Chart
plt.figure(figsize=(8, 6))
plt.bar(algorithms, speedup_vs_full, color=["royalblue", "darkorange", "mediumseagreen"])
plt.title("Speedup vs Full Search")
plt.ylabel("Speedup (x)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
